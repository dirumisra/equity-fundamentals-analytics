**# Missing value overview table created.**

```python
# Create an overview of missing values for each column
missing_overview = (
    raw_equity_df
    # Step 1: Identify missing values (True for NaN, False otherwise)
    .isna()
    
    # Step 2: Count missing values per column
    .sum()
    
    # Step 3: Convert the missing counts into a DataFrame
    .to_frame(name="missing_count")
    
    # Step 4: Add a new column 'missing_pct' to calculate % missing
    .assign(
        missing_pct=lambda x: (
            (x["missing_count"] / len(raw_equity_df)) * 100
        ).round(2)
    )
    
    # Step 5: Keep only columns that actually have missing values
    .query("missing_count > 0")
    
    # Step 6: Sort by missing percentage (highest first)
    .sort_values("missing_pct", ascending=False)
)

# Display the missing values overview table
missing_overview
```

#### **Data Cleaning Strategy (Design Phase)**

#### **1️⃣ Core Principle**

- Cleaning will be business-aware, conservative, and reversible.
- No column will be altered without a clear financial justification.

#### **2️⃣ Column-wise Cleaning Strategy**
🔹 Identifier Columns

**Columns: S.No., Name**

 - Must never be null
 - No filling, no transformation
 - Used only for identification
 - Any null here would be treated as a data quality error

**Action:**

➡ Keep as-is

➡ Validate only

---

#### **🔹 Categorical Columns**

**Columns:** Industries

 - Used for grouping and segmentation
 - Missing values may indicate unclassified companies

**Action:**

➡ Keep missing as NaN

➡ No imputation at this stage

➡ Consider “Unknown” only if required later (EPIC 4)

---

#### **🔹 Numeric Raw Columns (Absolute Business Facts)**

**Columns:**
CMP Rs., Mar Cap Rs.Cr., Sales Rs.Cr., Sales Qtr Rs.Cr.,
PAT 12M Rs.Cr., EPS 12M Rs.

- Represent absolute financial values
- Missing values usually due to reporting lag

**Action:**

➡ Do NOT fill with mean/median

➡ Keep missing as NaN

➡ Exclude rows only when metric is required (conditional filtering)

---

#### **3️⃣ Column Name Standardization Rules (Preview)**

- Convert to lowercase
- Replace spaces with _
- Replace % with _pct
- Remove special characters
- No semantic changes

(Actual renaming will happen in Task 2.2)

---

#### **4️⃣ Validation Rules (Post-cleaning)**

 - Row count must remain unchanged (unless explicitly filtered)
 - No identifier column becomes null
 - Numeric columns remain numeric
 - Derived metrics retain original meaning

 ---

#### **🚫 Explicitly Out of Scope (Task 2.1)**

 - No missing value filling
 - No column renaming
 - No row dropping
 - No recalculations

#### **Create Column Rename Mapping**

```python
# -------------------------------------------
# Column Rename Mapping
# -------------------------------------------

column_rename_map = {
    "S.No.": "s_no",
    "Industries": "industry",
    "Name": "company_name",
    "CMP Rs.": "cmp_rs",
    "Mar Cap Rs.Cr.": "market_cap_cr",
    "Sales Rs.Cr.": "sales_cr",
    "Sales Qtr Rs.Cr.": "sales_qtr_cr",
    "Sales growth %": "sales_growth_pct",
    "PAT 12M Rs.Cr.": "pat_12m_cr",
    "Profit growth %": "profit_growth_pct",
    "Profit Var 5Yrs %": "profit_var_5yrs_pct",
    "ROE %": "roe_pct",
    "ROCE %": "roce_pct",
    "OPM %": "opm_pct",
    "EPS 12M Rs.": "eps_12m_rs",
    "Debt / Eq": "debt_to_eq",
    "PEG": "peg",
    "P/E": "pe",
    "CMP / BV": "cmp_to_bv"
}

column_rename_map
```

#### **Apply column renaming (safe copy)**

- Deep copy: Changes made to the new "copy" won't affect the original data (since it's completely independent).
- Shallow copy: Changes made to the new "copy" will affect the original data too (since it's just a reference).

```python
# Rename the columns of the dataframe using the provided mapping and create a deep copy
equity_df = raw_equity_df.rename(columns=column_rename_map).copy()  

# Return the updated dataframe
equity_df.columns.tolist()
```

#### **Missing value baseline after column renaming**

```python
# Count missing values per column
missing_counts_post_rename = equity_df.isna().sum()

# Convert to DataFrame for readability
# Convert 'missing_counts_post_rename' into a DataFrame with a new column 'missing_count'
missing_summary_post_rename = (
    missing_counts_post_rename
    .to_frame(name="missing_count")  # Convert series to DataFrame and rename the column
    .assign(
        # Calculate the percentage of missing data for each entry, round to 2 decimal places
        missing_pct=lambda x: (x["missing_count"] / len(equity_df) * 100).round(2)
    )
    .query("missing_count > 0")  # Filter rows where there are missing values (i.e., 'missing_count' > 0)
    .sort_values("missing_pct", ascending=False)  # Sort the rows by 'missing_pct' in descending order
)

# Display the final DataFrame with the missing counts and their corresponding percentages
missing_summary_post_rename
```

#### **Normalize column headers (remove hidden spaces)**

```python
# Replace non-breaking spaces (represented as '\xa0') in column names with a regular space
equity_df.columns = equity_df.columns.str.replace("\xa0", " ", regex=False)

# Remove any leading or trailing whitespace characters from each column name
equity_df.columns = equity_df.columns.str.strip()

# Convert the cleaned column names to a list and return it
equity_df.columns.tolist()
```
